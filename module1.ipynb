## Preprocessing 

```python
df_2.duplicated().sum()
```

```python
df_2=df_2.drop_duplicates()
```

```python
df_2.isna().sum()
```

```python
from sklearn.model_selection import train_test_split

x=df_2.drop(['Class'], axis=1)
y=df_2['Class']

x_train,x_test,y_train,y_test=train_test_split(x,y,random_state=42,train_size=0.8,stratify=y)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler,FunctionTransformer
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer

time_28v_pipeline= Pipeline([
    ('imputer',SimpleImputer(strategy='mean')),
    ('scaler',StandardScaler())
])


amount_pipeline=Pipeline([

('imputer',SimpleImputer(strategy='median') ),
('log_transform',FunctionTransformer(np.log1p)),
('scaler',StandardScaler())
])

preprocessor=ColumnTransformer([
    ('time_28v',time_28v_pipeline, [col for col in x.columns if col !='Amount' ]),
    ('amount',amount_pipeline,['Amount'])
])
```

```python
x_train_processed=preprocessor.fit_transform(x_train)
x_test_processed=preprocessor.transform(x_test)
```

## Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
log_reg=LogisticRegression(C=1,class_weight='balanced',random_state=42,n_jobs=-1)
log_reg.fit(x_train_processed,y_train)

```

```python
from sklearn.metrics import precision_score,recall_score,f1_score,precision_recall_curve
from sklearn.model_selection import cross_val_predict

y_valid_predict=cross_val_predict(log_reg,x_train_processed,y_train,cv=5,n_jobs=-1,method='predict')

print('precision',precision_score(y_train,y_valid_predict))
print('recall',recall_score(y_train,y_valid_predict))
print('f1-score',f1_score(y_train,y_valid_predict))

```

## Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf=RandomForestClassifier(max_depth=20,random_state=42,n_jobs=-1,class_weight='balanced')
rf.fit(X=x_train_processed,y=y_train)
```

```python
from sklearn.metrics import precision_score,recall_score,f1_score,precision_recall_curve
from sklearn.model_selection import cross_val_predict,GridSearchCV

y_valid_predict=cross_val_predict(rf,x_train_processed,y_train,cv=5,n_jobs=-1,method='predict')

print('precision',precision_score(y_train,y_valid_predict))
print('recall',recall_score(y_train,y_valid_predict))
print('f1-score',f1_score(y_train,y_valid_predict))

```

## Fine tuning 

```python
param_grid={
'rf__n_estimators':[100,300],
'rf__max_depth':[5,10,15],
'rf__min_samples_split':[2,4,10],
'rf__min_samples_leaf':[2,4]
}
rf_pipe=Pipeline([
    ('preprocessor',preprocessor),
    ('rf', RandomForestClassifier(random_state=42,class_weight='balanced'))
])

rf_grid=GridSearchCV(estimator=rf_pipe,param_grid=param_grid,cv=3,verbose=1,n_jobs=-1,scoring='f1')
rf_grid.fit(x_train,y_train)
```

```python
rf_grid.best_params_
```

```python
rf=RandomForestClassifier(n_estimators=300,max_depth=10,random_state=42,min_samples_leaf=4,min_samples_split=10,n_jobs=-1,class_weight='balanced')
rf.fit(x_train_processed,y_train)
```

```python
rf_grid.best_score_
```

```python
from sklearn.metrics import average_precision_score

y_valid_reg_predict_proba=cross_val_predict(log_reg,x_train_processed,y_train,cv=5,n_jobs=-1,method='predict_proba')[:,1]
precision_log,recall_log,threshold_log=precision_recall_curve(y_train,y_valid_reg_predict_proba)
log_auc=average_precision_score(y_train,y_valid_reg_predict_proba)

y_valid_rf_predict_proba=cross_val_predict(rf,x_train_processed,y_train,cv=5,n_jobs=-1,method='predict_proba')[:,1]
precision_rf,recall_rf,threshold_rf=precision_recall_curve(y_train,y_valid_reg_predict_proba)
rf_auc=average_precision_score(y_train,y_valid_reg_predict_proba)
```

```python
plt.plot(recall_log[:-1], precision_log[:-1],
         label=f'logistic regression (AUC = {log_auc:.2f})')

plt.plot(recall_rf[:-1], precision_rf[:-1],
         label=f'random forest (AUC = {rf_auc:.2f})')

plt.xlabel('recall')
plt.ylabel('precision')
plt.title('precision recall curve')
plt.legend()
plt.show()
```
